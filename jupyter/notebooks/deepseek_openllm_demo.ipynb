# DeepSeek R1 OpenLLM Demo

This notebook demonstrates how to use the DeepSeek R1 model through OpenLLM for various text generation tasks.

## Setup

First, make sure you have OpenLLM installed and the DeepSeek model downloaded. This should already be done in your environment.

```python
import subprocess
import json
import time
import requests
from threading import Thread
import sys
import os

print("Setting up DeepSeek R1 model with OpenLLM...")
```

## Starting the OpenLLM Server

We'll start the DeepSeek model as a server that provides an OpenAI-compatible API.

```python
# Function to start OpenLLM server in background
def start_openllm_server():
    """Start OpenLLM server with DeepSeek model"""
    try:
        # Start server in background
        process = subprocess.Popen(
            ["openllm", "serve", "deepseek:r1-llama3.1-8b", "--port", "3000"],
            stdout=subprocess.PIPE,
            stderr=subprocess.PIPE,
            text=True
        )
        return process
    except Exception as e:
        print(f"Error starting OpenLLM server: {e}")
        return None

# Start the server
print("Starting OpenLLM server with DeepSeek R1 model...")
print("This may take a few minutes to download and load the model...")

# Note: In a real scenario, you might want to start this manually in a separate terminal
# For demo purposes, we'll show how to interact with it assuming it's running
```

## Interacting with DeepSeek R1

Once the server is running, we can interact with it using standard HTTP requests to the OpenAI-compatible API.

```python
import requests
import json

def query_deepseek(prompt, max_tokens=500, temperature=0.7):
    """Send a query to the DeepSeek model via OpenLLM API"""
    
    url = "http://localhost:3000/v1/chat/completions"
    
    payload = {
        "model": "deepseek:r1-llama3.1-8b",
        "messages": [
            {"role": "user", "content": prompt}
        ],
        "max_tokens": max_tokens,
        "temperature": temperature
    }
    
    headers = {
        "Content-Type": "application/json"
    }
    
    try:
        response = requests.post(url, json=payload, headers=headers, timeout=30)
        response.raise_for_status()
        return response.json()
    except requests.exceptions.RequestException as e:
        return {"error": f"Request failed: {e}"}
    except json.JSONDecodeError as e:
        return {"error": f"JSON decode error: {e}"}

def check_server_status():
    """Check if OpenLLM server is running"""
    try:
        response = requests.get("http://localhost:3000/v1/models", timeout=5)
        return response.status_code == 200
    except:
        return False

# Check server status
if check_server_status():
    print("✅ OpenLLM server is running and accessible!")
else:
    print("❌ OpenLLM server is not accessible. Please start it manually with:")
    print("   openllm serve deepseek:r1-llama3.1-8b --port 3000")
```

## Demo 1: Basic Text Generation

Let's start with a simple text generation example.

```python
# Basic text generation
prompt = "Explain the concept of machine learning in simple terms."

print(f"Prompt: {prompt}\n")
print("Generating response...\n")

response = query_deepseek(prompt)

if "error" in response:
    print(f"Error: {response['error']}")
else:
    answer = response['choices'][0]['message']['content']
    print(f"DeepSeek R1 Response:\n{answer}")
```

## Demo 2: Code Generation

DeepSeek models are particularly good at code generation. Let's test this capability.

```python
# Code generation example
code_prompt = """Write a Python function that takes a list of numbers and returns the median value. Include proper error handling and documentation."""

print(f"Code Generation Prompt: {code_prompt}\n")
print("Generating code...\n")

response = query_deepseek(code_prompt, max_tokens=800)

if "error" in response:
    print(f"Error: {response['error']}")
else:
    code = response['choices'][0]['message']['content']
    print(f"Generated Code:\n{code}")
```

## Interactive Demo

Here's a simple interactive function you can use to chat with DeepSeek:

```python
def chat_with_deepseek():
    """Interactive chat function with DeepSeek"""
    print("DeepSeek R1 Chat Interface")
    print("Type 'quit' to exit\n")
    
    while True:
        user_input = input("You: ")
        
        if user_input.lower() in ['quit', 'exit', 'bye']:
            print("Goodbye!")
            break
            
        if not user_input.strip():
            continue
            
        print("DeepSeek: Thinking...")
        response = query_deepseek(user_input)
        
        if "error" in response:
            print(f"Error: {response['error']}")
        else:
            answer = response['choices'][0]['message']['content']
            print(f"DeepSeek: {answer}\n")

# Uncomment the line below to start interactive chat
# chat_with_deepseek()
```

## Performance Monitoring

Let's add some basic performance monitoring to see how fast the model responds:

```python
import time

def timed_query(prompt, max_tokens=500):
    """Query DeepSeek with timing information"""
    start_time = time.time()
    response = query_deepseek(prompt, max_tokens=max_tokens)
    end_time = time.time()
    
    response_time = end_time - start_time
    
    if "error" not in response:
        content = response['choices'][0]['message']['content']
        word_count = len(content.split())
        words_per_second = word_count / response_time if response_time > 0 else 0
        
        print(f"Response time: {response_time:.2f} seconds")
        print(f"Words generated: {word_count}")
        print(f"Words per second: {words_per_second:.2f}")
        print(f"Content: {content}")
    else:
        print(f"Error: {response['error']}")

# Performance test
test_prompt = "What are the key benefits of using large language models in data analysis?"
print(f"Performance test with prompt: {test_prompt}\n")
timed_query(test_prompt)
```
