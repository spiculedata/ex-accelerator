# Sentiment Analysis with NLTK and TextBlob

This notebook demonstrates sentiment analysis techniques for social media text data.

## Features:
- TextBlob sentiment analysis
- NLTK VADER sentiment analyzer
- Visualization of sentiment distributions
- Comparison of different sentiment analysis methods

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from textblob import TextBlob
from nltk.sentiment import SentimentIntensityAnalyzer
import nltk

# Ensure NLTK data is available
try:
    nltk.data.find('vader_lexicon')
except LookupError:
    nltk.download('vader_lexicon')

# Initialize sentiment analyzer
sia = SentimentIntensityAnalyzer()

print('Libraries loaded successfully!')
print('Sentiment analyzers initialized!')
```

```python
# Sample social media texts
sample_texts = [
    "I absolutely love this new product! It's amazing! 😍",
    "This is terrible, worst experience ever. Completely disappointed.",
    "It's okay, nothing special but not bad either.",
    "Fantastic quality and great customer service! Highly recommend! 👍",
    "Not worth the money, very disappointed with the results.",
    "Mixed feelings about this. Some parts are good, others not so much.",
    "Outstanding performance! Exceeded all my expectations! 🎉",
    "Meh... it's just average. Nothing to get excited about."
]

# Create DataFrame
df = pd.DataFrame({'text': sample_texts})
print(f'Loaded {len(df)} sample texts')
df.head()
```

```python
# Perform sentiment analysis with TextBlob
def analyze_sentiment_textblob(text):
    blob = TextBlob(text)
    return {
        'polarity': blob.sentiment.polarity,
        'subjectivity': blob.sentiment.subjectivity
    }

# Perform sentiment analysis with VADER
def analyze_sentiment_vader(text):
    scores = sia.polarity_scores(text)
    return scores

# Apply sentiment analysis
df['textblob'] = df['text'].apply(analyze_sentiment_textblob)
df['vader'] = df['text'].apply(analyze_sentiment_vader)

# Extract TextBlob scores
df['tb_polarity'] = df['textblob'].apply(lambda x: x['polarity'])
df['tb_subjectivity'] = df['textblob'].apply(lambda x: x['subjectivity'])

# Extract VADER scores
df['vader_compound'] = df['vader'].apply(lambda x: x['compound'])
df['vader_pos'] = df['vader'].apply(lambda x: x['pos'])
df['vader_neu'] = df['vader'].apply(lambda x: x['neu'])
df['vader_neg'] = df['vader'].apply(lambda x: x['neg'])

print('Sentiment analysis completed!')
df[['text', 'tb_polarity', 'vader_compound']].head()
```
