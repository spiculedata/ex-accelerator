# Ollama Gemma3 Demo

This notebook demonstrates how to use the Gemma3 model through Ollama for various text generation tasks.

## Setup

Ensure the Ollama container is running and the `OLLAMA_HOST` environment variable is set.
The default model is `gemma4` which works well on modest hardware.

```python
import os
import ollama
import json
import time

print("Setting up Ollama Gemma3 client...")

# Create Ollama client using OLLAMA_HOST env var for containerised setup
client = ollama.Client(host=os.environ.get('OLLAMA_HOST', 'http://ollama:11434'))
DEFAULT_MODEL = "gemma4"
```

## Connecting to Ollama

We connect to Ollama using the `OLLAMA_HOST` environment variable, which defaults to `http://ollama:11434` for containerised deployments.

```python
# Verify Ollama connection and model availability
try:
    models = client.list()
    print("Ollama is running!")
    print(f"Available models: {len(models['models'])}")
    for model in models['models']:
        print(f"  - {model['model']}")
except Exception as e:
    print(f"Could not connect to Ollama: {e}")
    print(f"OLLAMA_HOST = {os.environ.get('OLLAMA_HOST', 'not set')}")
```

## Interacting with Gemma3

We use the Ollama Python client to chat with the model directly.

```python
def query_model(prompt, max_tokens=500, temperature=0.7):
    """Send a query to the Gemma3 model via Ollama"""
    try:
        response = client.chat(
            model=DEFAULT_MODEL,
            messages=[{'role': 'user', 'content': prompt}],
            options={
                'temperature': temperature,
                'num_predict': max_tokens,
            }
        )
        return response
    except Exception as e:
        return {"error": f"Request failed: {e}"}

def check_server_status():
    """Check if Ollama server is running"""
    try:
        client.list()
        return True
    except:
        return False

# Check server status
if check_server_status():
    print(f"Ollama server is running! Using model: {DEFAULT_MODEL}")
else:
    print("Ollama server is not accessible.")
    print("Ensure the Ollama container is running and OLLAMA_HOST is set.")
```

## Demo 1: Basic Text Generation

Let's start with a simple text generation example.

```python
# Basic text generation
prompt = "Explain the concept of machine learning in simple terms."

print(f"Prompt: {prompt}\n")
print("Generating response...\n")

response = query_model(prompt)

if "error" in response:
    print(f"Error: {response['error']}")
else:
    answer = response['message']['content']
    print(f"Gemma3 Response:\n{answer}")
```

## Demo 2: Code Generation

Let's test the model's code generation capability.

```python
# Code generation example
code_prompt = """Write a Python function that takes a list of numbers and returns the median value. Include proper error handling and documentation."""

print(f"Code Generation Prompt: {code_prompt}\n")
print("Generating code...\n")

response = query_model(code_prompt, max_tokens=800)

if "error" in response:
    print(f"Error: {response['error']}")
else:
    code = response['message']['content']
    print(f"Generated Code:\n{code}")
```

## Demo 3: Data Analysis Task

Let's ask Gemma3 to help with a data analysis scenario relevant to our ETL project.

```python
# Data analysis prompt
analysis_prompt = """I have a dataset of news articles with the following columns: title, content, published_date, source, and sentiment_score.
What are some interesting analytical questions I could explore with this data, and what Python libraries would be most suitable for each type of analysis?"""

print(f"Data Analysis Prompt: {analysis_prompt}\n")
print("Generating analysis suggestions...\n")

response = query_model(analysis_prompt, max_tokens=1000)

if "error" in response:
    print(f"Error: {response['error']}")
else:
    analysis = response['message']['content']
    print(f"Analysis Suggestions:\n{analysis}")
```

## Demo 4: Creative Writing

Let's test Gemma3's creative capabilities.

```python
# Creative writing prompt
creative_prompt = """Write a short story (2-3 paragraphs) about a data scientist who discovers something unexpected in a dataset about global news patterns."""

print(f"Creative Writing Prompt: {creative_prompt}\n")
print("Generating story...\n")

response = query_model(creative_prompt, max_tokens=800, temperature=0.9)

if "error" in response:
    print(f"Error: {response['error']}")
else:
    story = response['message']['content']
    print(f"Generated Story:\n{story}")
```

## Demo 5: Problem Solving

Let's test Gemma3's reasoning capabilities with a logic problem.

```python
# Problem solving prompt
problem_prompt = """You have 3 boxes. One contains only apples, one contains only oranges, and one contains both apples and oranges.
All boxes are labeled incorrectly. You can pick one fruit from one box to determine the contents of all boxes.
Which box should you pick from and why?"""

print(f"Problem Solving Prompt: {problem_prompt}\n")
print("Generating solution...\n")

response = query_model(problem_prompt, max_tokens=600)

if "error" in response:
    print(f"Error: {response['error']}")
else:
    solution = response['message']['content']
    print(f"Solution:\n{solution}")
```

## Interactive Demo

Here's a simple interactive function you can use to chat with Gemma3:

```python
def chat_with_model():
    """Interactive chat function with Gemma3"""
    print("Gemma3 Chat Interface")
    print("Type 'quit' to exit\n")
    
    while True:
        user_input = input("You: ")
        
        if user_input.lower() in ['quit', 'exit', 'bye']:
            print("Goodbye!")
            break
            
        if not user_input.strip():
            continue
            
        print("Gemma3: Thinking...")
        response = query_model(user_input)
        
        if "error" in response:
            print(f"Error: {response['error']}")
        else:
            answer = response['message']['content']
            print(f"Gemma3: {answer}\n")

# Uncomment the line below to start interactive chat
# chat_with_model()
```

## Performance Monitoring

Let's add some basic performance monitoring to see how fast the model responds:

```python
import time

def timed_query(prompt, max_tokens=500):
    """Query Gemma3 with timing information"""
    start_time = time.time()
    response = query_model(prompt, max_tokens=max_tokens)
    end_time = time.time()
    
    response_time = end_time - start_time
    
    if "error" not in response:
        content = response['message']['content']
        word_count = len(content.split())
        words_per_second = word_count / response_time if response_time > 0 else 0
        
        print(f"Response time: {response_time:.2f} seconds")
        print(f"Words generated: {word_count}")
        print(f"Words per second: {words_per_second:.2f}")
        print(f"Content: {content}")
    else:
        print(f"Error: {response['error']}")

# Performance test
test_prompt = "What are the key benefits of using large language models in data analysis?"
print(f"Performance test with prompt: {test_prompt}\n")
timed_query(test_prompt)
```
